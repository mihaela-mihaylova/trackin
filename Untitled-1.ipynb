```python
import pandas as pd
import os

# cleans the raw trackmate file as saved from trackmate
def tidy_up_trackmate_tracks(raw_trackmate_df):
    
    # drop first three rows, as they are simply extra column names that are unnecessary
    trackmate_df = raw_trackmate_df.drop([0, 1, 2])

    # Reset the index after dropping rows
    trackmate_df = trackmate_df.reset_index(drop=True)
    trackmate_df = trackmate_df[['FRAME', 'POSITION_Y', 'POSITION_X', 'TRACK_ID']]
    trackmate_df.columns = ['tframe', 'y', 'x', 'track_no']
        
    # round up the pixel values
    trackmate_df['y'] = round(pd.to_numeric(trackmate_df['y']))
    trackmate_df['x'] = round(pd.to_numeric(trackmate_df['x']))

    return trackmate_df


raw_trackmate_df = pd.read_csv('/Users/mmihaylova/Downloads/videos_to_redo/26_neuts-221125-LifeAct-nTnG-Dock8-KO-6um/221125_advanced_kalman.csv')
trackmate_df = tidy_up_trackmate_tracks(raw_trackmate_df)
trackmate_df.to_csv('/Users/mmihaylova/Downloads/videos_to_redo/26_neuts-221125-LifeAct-nTnG-Dock8-KO-6um/221125_cleaned_advanced_kalman.csv', index=False)

           
        
```

```python
import pandas as pd
import os

# Load the dataframe into pandas
file_path = "/Users/mmihaylova/Downloads/channels_20220524_LifeAct_nTnG_30um.csv"
directory_name = os.path.dirname(file_path)

df = pd.read_csv(file_path)

# Get the base file name without extension
base_filename = os.path.basename(file_path).replace(".csv", "")

# Iterate over each unique `track_no` in the dataframe
for track_no in df['channel'].unique():
    # Skip NaN values using pd.notna()
    if pd.notna(track_no):
        # Filter rows corresponding to the current `track_no`
        track_df = df[df['channel'] == track_no]
    
        # Create a new filename with the format `{i}_original_file_name.csv`
        # Use string formatting with track_no directly
        output_filename = f"{int(track_no)}_{base_filename}.csv"
        
        # Save the filtered dataframe to a new CSV file in the same directory
        track_df = track_df[['tframe','y','x','track_no']]
        track_df.to_csv(os.path.join(directory_name, output_filename), index=False)
        print(f"Saved: {output_filename}")
```
